**2. In a markdown block, outline why more than 3 degrees of polynomial expression are not recommended in Machine Learning.**


Using more then 3 degrees of polynomial expression isn't recommended in machine learning since anything above 3 degrees is considered as "over fitting" the model. Over fitting the model is when the line tries to pass through most of if not all the points on the graph, leading to exceptional accuracy when presented with the training data, but large errors to data outside of what it has been trained on since it becomes very sensitive to even the smallest of outliers.
