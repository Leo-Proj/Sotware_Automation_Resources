# Iris Logistic Regression Implementation

The modified Iris flower data set consists of 50 samples from each of two species of Iris (Iris setosa and Iris versicolor) for binary classification. Four features were measured from each sample: the length and the width of the sepals and petals, in centimeters. *In week 4 you will apply a Neural Network to the full Iris dataset for multiway classification*.

Your task is to design, fit and evaluate a Logistic Regression model to clasify the each species.

| Binary | Class |
| --- | --- |
| 0 | versicolor |
| 1 | setosa |

#### Step 1 - Import Dependencies

```python
import numpy as np
import pandas as pd

%matplotlib inline
import matplotlib.pyplot as plt
plt.style.use('ggplot')

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import mean_absolute_error,mean_squared_error 
```

#### Step 2 - Import And Visualise The Data

```python
# Import data as a Pandas DataFrame
iris = pd.read_csv('data/iris_modified.csv', sep='\t')
```

```python
# Plot the data
fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Plot all combinations of features
axes[0, 0].scatter(
    iris["sepal_length"], iris["sepal_width"], c=iris["species"], cmap="viridis"
)
axes[0, 0].set_xlabel("Sepal Length")
axes[0, 0].set_ylabel("Sepal Width")

axes[0, 1].scatter(
    iris["sepal_length"], iris["petal_length"], c=iris["species"], cmap="viridis"
)
axes[0, 1].set_xlabel("Sepal Length")
axes[0, 1].set_ylabel("Petal Length")

axes[0, 2].scatter(
    iris["sepal_length"], iris["petal_width"], c=iris["species"], cmap="viridis"
)
axes[0, 2].set_xlabel("Sepal Length")
axes[0, 2].set_ylabel("Petal Width")

axes[1, 0].scatter(
    iris["sepal_width"], iris["petal_length"], c=iris["species"], cmap="viridis"
)
axes[1, 0].set_xlabel("Sepal Width")
axes[1, 0].set_ylabel("Petal Length")

axes[1, 1].scatter(
    iris["sepal_width"], iris["petal_width"], c=iris["species"], cmap="viridis"
)
axes[1, 1].set_xlabel("Sepal Width")
axes[1, 1].set_ylabel("Petal Width")

axes[1, 2].scatter(
    iris["petal_length"], iris["petal_width"], c=iris["species"], cmap="viridis"
)
axes[1, 2].set_xlabel("Petal Length")
axes[1, 2].set_ylabel("Petal Width")

plt.tight_layout()
plt.show()
```

```python
# Store the strongest features and target categories for using in the model
X = iris[["petal_length", "petal_width"]].values
y = iris["species"].values
```

#### Step 3: Instantiate a LogisticRegression Object

```python
model = LogisticRegression()
```

#### Step 4: Fit the model

```python
model.fit(X, y)
```

Now that the regression has been fit, we can use the `predict` method to forecast whether our model thinks a family will switch wells.

```python
# Predict the target for the whole dataset
y_pred = model.predict(X)
print("Predictions:", y_pred[:10])
print("Actual:", y[:10])
```

Additionally (and in many cases more usefully), the `LogisticRegression` has a `predict_proba` method that produces a predicted **probability**.

```python
# Predict the probabilities for the whole dataset
y_pred_proba = model.predict_proba(X)

# Print the predicted probabilities
print("Predicted probabilities (first 5 samples):")
print(y_pred_proba[:5])
```

#### Step 5: Plot the Predictions

```python
# Plot the predictions compared to the actual data
# Step 5: Plot the predictions compared to the actual data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c=y, cmap="viridis", edgecolors="k", s=100)
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Actual Species")
plt.colorbar(label="Species")

plt.subplot(1, 2, 2)
plt.scatter(X[:, 0], X[:, 1], c=y_pred, cmap="viridis", edgecolors="k", s=100)
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Predicted Species")
plt.colorbar(label="Species")

plt.tight_layout()
plt.show()
```

#### Step 6: Evaluate The Model

The model's prediction shows a near perfect parallel to its training data, showing that the logistic regression model is very good at classifying the different types of flowers.
