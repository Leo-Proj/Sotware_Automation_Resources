__7. Refering to the below model, describe underfitting in Linear Regression.__

````python
x_train = np.array([1, 1.5, 1.8, 2, 2.2])
y_train = np.array([[200], [250], [260], [280], [500]])
tmp_f_wb = compute_model_output(x_train, 200, 10,)
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')
plt.show()
````

In linear regression, underfitting is when the graph's prediction line is too far away from the actual values, making the machine learning model too simple and inaccurate. In the model above, we can see how the majority of the red x's (representing actual values) of the training set data are too far away from the prediction line which will result in inaccurate predictions.
